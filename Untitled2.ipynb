```python
#6
gdp = gdp.rename(columns={'Country': 'country'})
literacy_rate = literacy_rate.rename(columns={'name': 'country'})
```

```python
#8
merged_df = population.merge(gdp, on='country', how='inner')
merged_df = merged_df.merge(internet_users, on='country', how='inner')
merged_df = merged_df.merge(literacy_rate, on='country', how='inner')

merged_df.head()
```

```python
#9
merged_df['Internet Penetration (%)'] = (merged_df['internet_users'] / merged_df['population']) * 100
merged_df['GDP Per Capita'] = merged_df['GDP'] / merged_df['population']

merged_df.head()
```

```python
#10

import sqlite3

conn = sqlite3.connect('country_statistics.db')
merged_df.to_sql('country_statistics', conn)

```

```python
#11
q = "SELECT * FROM country_statistics LIMIT 5"
db = pd.read_sql_query(q, conn)

db
```
