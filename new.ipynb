```python
#3. Password Validator
#Write a function that checks if a password is strong: at least 8 characters,
#contains at least one digit, one uppercase letter, and one special character
#(!@#$%).
password=input('enter the password')
upper=False
lower=False
digit=False
special=False
spl=('!@#$%')
for letter in password:
  if letter.isupper():
    upper = True
  elif letter.islower():
    lower = True
  elif letter.isdigit():
    digit = True
  elif letter in spl:
    special = True
if len(password) >= 8 and upper and lower and digit and special:
  print("Strong password")
else:
  print("Weak password")
```

```python
#4. Word Frequency Counter
#Given a paragraph of text, write a function that returns the 3 most
#frequently occurring words (ignore case and punctuation).
text = input("enter a word ")
letter_frequency= {}
for i in text:
  if i in letter_frequency:
    letter_frequency[i]+=1
  else:
    letter_frequency[i] = 1
print(letter_frequency)
```

```python
 #5.FizzBuzz Variant
#Print numbers 1 to 50, but replace multiples of 3 with "Fizz", multiples of 5
#with "Buzz", multiples of both with "FizzBuzz" — and multiples of 7 with
#"Bang".
for value in range(1,51):
  if value%3==0 and value%5==0:
    print("FizzBuzz")
  elif value%3==0:
    print("Fizz")
  elif value%5==0:
    print("Buzz")
  elif value%7==0:
    print("bang")
  else:
    print(value)
```
